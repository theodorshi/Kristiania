## Data and helper function

### City Data

Contains two dictionaries with key/value pairs.
cities: key = city name, values = latitude, longitude, size.
poi: key = latitude and longitude, value = poi name and rating

```python
cities = {'Bergen': [60.364098, 5.404422, 15],
          'Oslo': [59.926959, 10.779454, 10],
          'Trondheim': [63.433654, 10.390277, 5]
}
poi = {(63.426806, 10.396712): ['Nidarosdommen', 4.5],
        (59.926699, 10.701075): ['Vigelandsparken', 5],
        (59.907704, 10.753163): ['Operahuset', 4],
        (60.397561, 5.322833): ['Bryggen', 4.5],
        (60.387639, 5.321623): ['Universitetmuseet', 5],
        (60.342901, 5.336796): ['Gamlehaugen', 4]
        }
```

### Calculating distance

This function calculates the distance between two points. One longitude equals 68 km, and one latitude equals 111 km. The square root is used to calculate the direct distance between the coordinates.

```python
import math

def calculate_distance(lat1, lon1, lat2, lon2):
    # Calculate distance between two coordinates
    dLon = (lon1 - lon2) * 68
    dLat = (lat1 - lat2) * 111

    # Returns diagonal distance
    return math.sqrt(dLat**2 + dLon**2)
```

## 3.a) Functions that returns closest POI and city for a given pair of coordinates

### Closest POI

This function takes two coordinates as arguments, and calculates the distance to all POIs. The name of the POI with the shortest distance to the given coordinates is returned.

```python
def calculate_nearest_poi(lat, lon):
    poi_name = None
    shortest_distance = None

    # key = coordinates, values = info (name and rating)
    for coordinates, info in poi.items():
        p_lat, p_lon = coordinates # unpacking coordinates in POI

        distance = calculate_distance(p_lat, p_lon, lat, lon)

        if shortest_distance is None or distance < shortest_distance:
            shortest_distance = distance
            poi_name = info[0]

    return poi_name
```

### Closest City

This works the same way as the function above, but instead calculates the distance to all cities and returns the name of the closest city.

```python
def calculate_nearest_city(lat, lon):
    closest_city = None
    shortest_distance = None

    # Unpacks key and value, key = city name, value = array with coordinates and size
    for city_name, city_data in cities.items():
        city_lat = city_data[0]
        city_lon = city_data[1]

        distance = calculate_distance(city_lat, city_lon, lat, lon)

    # Checks which city is closest to given coordinates
        if shortest_distance is None or distance < shortest_distance:
            shortest_distance = distance
            closest_city = city_name

    return closest_city
```

### Function that prints nearest city and POI for a given pair of coordinates

showNearest prompts the user to enter latitude and longitude coordinates. These values are then passed to the functions that retrieve the closest poi and city and prints them out to the console.

```python
def show_nearest():
    # Userinput
    lat = float(input("Enter latitude: "))
    lon = float(input("Enter longitude: "))

    # Calculates nearest from given coordinates
    nearest_city = calculate_nearest_city(lat, lon)
    nearest_poi = calculate_nearest_poi(lat, lon)

    print("Nearest city is: ", nearest_city)
    print("Nearest poi is: ", nearest_poi)

show_nearest()
```

## 3.b) Compute city boundaries

For each city the coordinates for north, south, east and west boundaries are calculated based on the city size. These boundaries are then added to the cities dictionary as a tuple in the values.

```python
def add_boundaries():
    for city_name, city_data in cities.items():
        lat, lon, size = city_data

        # Calculates boundaries based on citysize
        north = round(lat + (size / 111), 6)
        south = round(lat - (size / 111), 6)
        east = round(lon + (size / 68), 6)
        west = round(lon - (size / 68), 6)

        # Overwriting dictionary to add boundaries
        cities[city_name] = [[lat, lon, size], (north, south, east, west)]

        # Updated city structure:
        # [
        #   [lat, lon, size],
        #   [north, south, east, west]
        # ]

    print(cities)

add_boundaries()
```

## 3.c) Assign POIs to cities

This function loops through all cities and compares their boundaries to the coordinates of all POIs. If the POI is located inside the city boundaries, it is added to a new array to store POIs for each city. The array is then added to the values of each city.

```python
def assign_poi_to_city():
    for city_name, city_data in cities.items():
        north, south, east, west = city_data[1]

        # New array to store POI
        pois_city = []

        for coordinates, info in poi.items():
            poi_lat, poi_lon = coordinates
            poi_name, poi_rating = info

            # Checks if POI is located inside city boundaries
            if south <= poi_lat <= north and west <= poi_lon <= east:
                pois_city.append([poi_lat, poi_lon, poi_name, poi_rating])

        # Overwriting dictionary and adding array with pois
        cities[city_name] = [city_data[0], city_data[1], pois_city]

        # Updated city structure:
        # [
        #   [lat, lon, size],
        #   [north, south, east, west],
        #   [pois]
        # ]
    print(cities)

assign_poi_to_city()
```

## 3.d) Compute average POI rating per city

In this function, the added POI array in cities is used to calculate the average rating of all POIs within each city.

```python
def get_poi_rating():
    for city_name, city_data in cities.items():
        pois_city = city_data[2]

        total_rating = 0

        for p in pois_city:
            rating = p[3] # rating is at index 3 in poi
            total_rating += rating # Adds up ratings for all POIs in a city

        average_rating = total_rating / len(pois_city) # Calculate avr rating

        print(city_name, average_rating)

get_poi_rating()
```

## 3.e) Order POIs by distance from the city center

This function loops through all cities and their POIs. The distance from the city center to each POI is then calculated, and the values is stored in an array. The array is then sorted in ascending order using the sort() method and printed.

```python
def sort_poi():
    for city_name, city_data in cities.items():
        poi_city = city_data[2]
        city_coordinates = city_data[0]
        city_lat = city_coordinates[0]
        city_lon = city_coordinates[1]

        # Array to store distance and name of POI
        poi_distance_center = []
        for p in poi_city:
            poi_lat, poi_lon, poi_name, poi_rating = p

            # Calculates distance form center and appends to array
            poi_distance = calculate_distance(city_lat, city_lon, poi_lat, poi_lon)
            poi_distance_center.append([poi_distance, poi_name])

        # Sorts on distance from center
        poi_distance_center.sort()
        print(city_name, poi_distance_center)

sort_poi()
```
