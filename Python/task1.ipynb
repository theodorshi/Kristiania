```python
def supernacci(n):
    s = [1] #Starting value list

    for i in range(2, n + 1): # Starting at index 2 since s(1) is set
        j = (i - 1) % 3 # Calculation
        if j == 0:
            next_val = s[-1] # Setting previous value in list
        else:
            next_val = sum(s[-j:]) # Summing value based on result from j. j is then the number of indexes we need to sum
        s.append(next_val) # Adding number to end of the list

    print(f"Index n: {n}")
    print(f"Sequence built: {s}")
    print(f"The {n}-th Supernacci number: {s[n - 1]}\n")

    return s[n - 1]  #returns only the n-th Supernacci number

#Tests with n = 6 and n = 4

print("Testing Supernacci")
test_result_1 = supernacci(6)
test_result_2 = supernacci(4)
```

### Explanation

The `supernacci(n)` function builds the Supernacci sequence iteratively:

- We start with the list `s = [1]`, representing `s(1) = 1` (the base case).
- The loop runs for each value `i` from 2 up to and including `n` – a total of **n − 1 iterations**.
- In each iteration, we compute `j = (i - 1) % 3`, which gives either 0, 1, or 2.
- If `j == 0` (the sum is empty), we use the previous value in the list as the next number.
- Otherwise, we sum the last `j` numbers in the list using slicing (`s[-j:]`) and `sum()`.
- We append the new value to the list with `.append()`.
- Finally, we return `s[n-1]` – the **n-th** Supernacci number.

This approach uses `sum()` and slicing as required by the task, and avoids recursion.
